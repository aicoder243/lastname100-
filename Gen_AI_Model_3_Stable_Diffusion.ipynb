```python
import torch
from diffusers import StableDiffusionPipeline
from IPython.display import display

# Use a GPU if available
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Running on:", device)

# Load a pretrained text-to-image model
model_id = "stable-diffusion-v1-5/stable-diffusion-v1-5"

image_generator = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16 if device == "cuda" else torch.float32
)

image_generator = image_generator.to(device)

print("Image-generation model loaded!")

#If the model repository requests access approval or authentication, follow the repository's current instructions. A GPU is strongly recommended; CPU execution may be very slow.
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Running on:", device)
```

```python
prompt = """
A modern reusable smart water bottle,
minimalist product design, white background,
professional commercial product photography,
realistic lighting, high detail
"""

# Generate the image
result = image_generator(
    prompt=prompt,
    num_inference_steps=25,
    guidance_scale=7.5
)

# Display the generated image
generated_image = result.images[0]
display(generated_image)
```
